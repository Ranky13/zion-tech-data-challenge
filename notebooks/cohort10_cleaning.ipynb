```python
import pandas as pd
import numpy as np


cohort10 = pd.read_excel('data/Cohort 10.0 Registration form (Responses).xlsx')

cohort_10_cleaned = cohort10.copy()

cohort_10_cleaned = cohort_10_cleaned.rename(columns={
    'Timestamp': 'Timestamp',
    'NAME': 'Name',
    'EMAIL ADDRESS': 'Email Address',
    'PHONE NUMBER  (Add country code e.g +234)': 'Phone Number',
    'COUNTRY  (e.g. Nigeria, UK)': 'Country',
    'Gender': 'Gender',
    'CHOICE OF COURSE': 'Course',
    'Occupation': 'Occupation',
    'How did you hear about our Program?': 'Referral Source'
})

cohort_10_cleaned['Day Time'] = pd.cut(
    cohort_10_cleaned['Timestamp'].dt.hour,
    bins=[0, 5, 12, 19, 24],
    labels=['Midnight', 'Morning', 'Afternoon', 'Night'],
    right=False,
    include_lowest=True
)


cohort_10_cleaned['Cohort'] = '10'
cohort_10_cleaned['Email Address'] = cohort_10_cleaned['Email Address'].astype(str).str.strip().str.lower()
cohort_10_cleaned['Name'] = cohort_10_cleaned['Name'].astype(str).str.strip().str.title()
cohort_10_cleaned['Phone Number'] = cohort_10_cleaned['Phone Number'].astype(str).str.replace(r'\D', '', regex=True)
cohort_10_cleaned['Country'] = cohort_10_cleaned['Country'].astype(str).str.strip().str.title()
cohort_10_cleaned['Gender'] = cohort_10_cleaned['Gender'].astype(str).str.strip().str.title()
cohort_10_cleaned['Course'] = cohort_10_cleaned['Course'].astype(str).str.strip().str.title()
cohort_10_cleaned['Occupation'] = cohort_10_cleaned['Occupation'].astype(str).str.strip().str.title()
cohort_10_cleaned['Occupation'] = cohort_10_cleaned['Occupation'].fillna('Not Specified')
cohort_10_cleaned['Referral Source'] = cohort_10_cleaned['Referral Source'].astype(str).str.strip().str.title()

cohort_10_cleaned.to_excel('data/cohort10_cleaned.xlsx', index=False)


```
