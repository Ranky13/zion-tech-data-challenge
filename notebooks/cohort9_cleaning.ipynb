```python
import pandas as pd
import numpy as np
np.set_printoptions(threshold=np.inf)

cohort9 = pd.read_excel('data/Cohort 9.0 (Responses).xlsx')


cohort_nine_cleaned = cohort9.copy()

cohort_nine_cleaned = cohort_nine_cleaned.rename(columns={
    'Timestamp': 'Timestamp',
    'NAME': 'Name',
    'Email address': 'Email Address',
    'Phone Number (Add country code e.g +234)': 'Phone Number',
    'Country (e.g. Nigeria, UK)': 'Country',
    'Gender': 'Gender',
    'Choice of Program': 'Course',
    'How did you hear about our Program?': 'Referral Source',
    })


cohort_nine_cleaned['Day Time'] = pd.cut(
    cohort_nine_cleaned['Timestamp'].dt.hour,
    bins=[0, 5, 12, 19, 24],
    labels=['Midnight', 'Morning', 'Afternoon', 'Night'],
    right=False,
    include_lowest=True
)
cohort_nine_cleaned['Cohort'] = '9'
cohort_nine_cleaned['Email Address'] = cohort_nine_cleaned['Email Address'].astype(str).str.strip().str.lower()
cohort_nine_cleaned['Name'] = cohort_nine_cleaned['Name'].astype(str).str.strip().str.title()
Email_name = cohort_nine_cleaned['Email Address'].str.split('@').str[0].str.title()
missing_name = cohort_nine_cleaned['Name'].isna()
cohort_nine_cleaned.loc[missing_name, 'Name'] = Email_name[missing_name]
cohort_nine_cleaned['name_from_email'] = missing_name
cohort_nine_cleaned['Phone Number'] = cohort_nine_cleaned['Phone Number'].astype(str).str.replace(r'\D', '', regex=True)
cohort_nine_cleaned['Country'] = cohort_nine_cleaned['Country'].astype(str).str.strip().str.title()
cohort_nine_cleaned['Gender'] = cohort_nine_cleaned['Gender'].astype(str).str.strip().str.title()
cohort_nine_cleaned['Referral Source'] = cohort_nine_cleaned['Referral Source'].astype(str).str.strip().str.title()

country_map = {
    'Nige': 'Nigeria',
    'Nigera': 'Nigeria',
    'Nigeria L': 'Nigeria',
    'Nigeria Nigeria': 'Nigeria',
    'Nigerian': 'Nigeria',
    'Niger': 'Nigeria',
    'Nigerianl': 'Nigeria',
    'Nig': 'Nigeria',
    'Nigeria Ekiti State': 'Nigeria',
    'Nigeria, Russia.': 'Nigeria',
    'Uk': 'United Kingdom',
    'Usa': 'United States',
    'Us': 'United States',
    'Ghan': 'Ghana',
    'Botswana Africa': 'Botswana',
    'South African': 'South Africa',
    'Rsa': 'South Africa',
    'Sa': 'South Africa',
    'Sout': 'South Africa',
    '+27 81 313 5587': 'South Africa',
    'Benin Republic': 'Benin',
    'Tanzania,Dar Es Salaam': 'Tanzania',
    'España': 'Spain',
    'Cote D Ivoire': 'Ivory Coast',
    'Belarus (Nigerian)': 'Unknown',
    'U': 'Unknown',
    '8 056 132997': 'Unknown',
    'E.G.': 'Unknown',
}

cohort_nine_cleaned['Country'] = cohort_nine_cleaned['Country'].replace(country_map)
cohort_nine_cleaned['Course'] = cohort_nine_cleaned['Course'].astype(str).str.strip().str.title()


cohort_nine_cleaned.to_excel('data/cohort9_cleaned.xlsx', index=False)
print(cohort_nine_cleaned.info())

```
